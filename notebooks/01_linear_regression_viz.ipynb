# Visualizing Linear Regression
This notebook explores the manual implementation of linear regression against synthetic data.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.fundamentals.linear_regression import LinearRegression

# Generate synthetic data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1).flatten()

# Fit model
model = LinearRegression(learning_rate=0.1, n_iterations=1000)
model.fit(X, y)

# Visualize
plt.scatter(X, y, color='blue', label='Data')
X_range = np.linspace(0, 2, 100).reshape(-1, 1)
y_pred = model.predict(X_range)
plt.plot(X_range, y_pred, color='red', label='Prediction')
plt.legend()
plt.title("Linear Regression Fit")
plt.show()
```
